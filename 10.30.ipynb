```python
def fib(n):
    a,b=0,1
    x=0
    s=[]
    while x<n:
        sum=a+b
        s.append(a)
        a,b=b,sum
        x=x+1
    return s


n=int(input("输入"))
print(fib(n))
        
```

```python
def fib(n):
    if n < 2:
        return n
    return fib(n - 1) + fib(n - 2)

# 取前 10 项
print([fib(i) for i in range(10)])
```
