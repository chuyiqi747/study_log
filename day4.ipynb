```python
# 过程式
squares = []
for x in range(5):
    squares.append(x*x)

# 声明式（读起来像英语：x 的平方，对 range(5) 里每个 x）
squares = [x*x for x in range(5)]
evens   = [x for x in range(20) if x % 2 == 0]   # 加 if 就是筛选
print(squares,evens)
```
