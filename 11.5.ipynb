```python

from collections import defaultdict
import csv
scores= defaultdict(list)
with open ("data/scores.csv", encoding="utf-8",newline="") as f:
    for row in csv.DictReader(f):
        scores[row["姓名"]].append(int(row["分数"]))
result = sorted(
((name,sum(ns)) for name,ns in scores.items()),key=lambda kv: kv[1], reverse=True

)
report = "成绩汇总（按sum降序）\n"
report += "\n".join(f"{name}: {sum:.1f}" for name, sum in result) + "\n"
with open("data/report.txt", "w", encoding="utf-8") as f:
    f.write(report)
    

        
    
    
```
