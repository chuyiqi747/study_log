```python
def sum_n(n):
    total = 0
    for i in range(1, n+1):
        total += i
    return total

print(sum_n(1))
```

```python
# """判断是否为质数"""
import math
def is_prime(n):
    if n<1:
        return False
    x=int(math.sqrt(n))
    for i in range(2,x+1):
        if n%i==0:
            print("no")
            return False
    print("yes")
    return True
    

n = int(input("输入数字："))
is_prime(n)
```
