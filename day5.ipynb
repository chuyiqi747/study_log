### import random

def make_secret(low=1, high=100):
    return random.randint(low, high)
def judge(guess, secret):
   
        if guess < secret:
            return "小了"
        elif guess > secret:
            return "大了"
        return "对了"
def play_game(low=1, high=100, max_tries=None):
    secret = make_secret(low, high)   # 叫出题机
    tries = 0
    print(f"我想好了一个 {low}~{high} 之间的整数，来猜吧！")
    
    while True:
            if max_tries is not None and tries >= max_tries:
                print(f"次数用完了～答案是 {secret}。")
                return False
    
            raw = input(f"第 {tries + 1} 次，你的猜测：")
    
            # EAFP：先假设能转成整数，转不成再收拾
            try:
                guess = int(raw)
            except ValueError:
                print("这好像不是数字哦，请输入整数～")
                continue
    
            tries += 1
            result = judge(guess, secret)   # 叫判定机
            if result == "对了":
                print(f"🎉 对了！就是 {secret}，你用了 {tries} 次。")
                return True
            print(result)


if __name__ == "__main__":
    play_game()

```python
def calc(a, op, b):
    match op:
        case "+": return a + b
        case "-": return a - b
        case "*": return a * b
        case "/":
            if b == 0:
                return "除零错误"
            return a / b
        case _:
            return "未知运算符"

def main():
    try:
        a = float(input("第一个数: "))
        op = input("运算符(+ - * /): ")
        b = float(input("第二个数: "))
    except ValueError:
        print("数字格式不对"); return
    print("结果:", calc(a, op, b))

if __name__ == "__main__":
    main()
```
